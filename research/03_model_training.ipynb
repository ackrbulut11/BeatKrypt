```python
import os
import numpy as np
import pandas as pd
from google.colab import drive

# PyTorch ve Derin Öğrenme Bileşenleri
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader

# Model Değerlendirme ve Veri Bölme (Scikit-Learn)
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score

# Görselleştirme ve Grafik Çizimi
import matplotlib.pyplot as plt
import seaborn as sns

# Google Drive bağlantısı
drive.mount('/content/drive')

# Dosya yolları ve temel ayarlar
drive_base_path = '/content/drive/MyDrive/BeatKrypt/'
OUTPUT_DIR = os.path.join(drive_base_path, 'processed_data_split')
```

```python
# --- HIGH QUALITY SETİ YÜKLENİYOR ---
X_high_left = np.load(os.path.join(OUTPUT_DIR, "high/X_high_left.npy"))
X_high_right = np.load(os.path.join(OUTPUT_DIR, "high/X_high_right.npy"))
y_high = np.load(os.path.join(OUTPUT_DIR, "high/y_high.npy"))
meta_high = pd.read_csv(os.path.join(OUTPUT_DIR, "high/metadata_high.csv"))

# --- MEDIUM QUALITY SETİ YÜKLENİYOR ---
X_med_left = np.load(os.path.join(OUTPUT_DIR, "medium/X_medium_left.npy"))
X_med_right = np.load(os.path.join(OUTPUT_DIR, "medium/X_medium_right.npy"))
y_med = np.load(os.path.join(OUTPUT_DIR, "medium/y_medium.npy"))
meta_med = pd.read_csv(os.path.join(OUTPUT_DIR, "medium/metadata_medium.csv"))

print(f"✓ High Quality Kümesi Yüklendi: {len(y_high)} çift")
print(f"✓ Medium Quality Kümesi Yüklendi: {len(y_med)} çift")
```

```python
# NumPy Array'lerini dikey olarak uç uca ekliyoruz (Concat)
X_all_left = np.concatenate([X_high_left, X_med_left], axis=0)
X_all_right = np.concatenate([X_high_right, X_med_right], axis=0)
y_all = np.concatenate([y_high, y_med], axis=0)

# Metadata DataFrame'lerini Pandas ile birleştiriyoruz
metadata_all = pd.concat([meta_high, meta_med], ignore_index=True)

print("=== BİRLEŞTİRİLMİŞ NİHAİ VERİ HAVUZU RAPORU ===")
print(f"Toplam Sol Sinyal (X_all_left) shape  : {X_all_left.shape}")
print(f"Toplam Sağ Sinyal (X_all_right) shape : {X_all_right.shape}")
print(f"Toplam Etiket (y_all) shape           : {y_all.shape}")
print(f"Toplam Metadata Satır Sayısı          : {len(metadata_all)}")

print(f"\nSınıf Dağılımı (0: Farklı Kişi, 1: Aynı Kişi):")
print(pd.Series(y_all).value_counts())
```
